```python
import pandas as pd 
import numpy as np 
```

```python
# import csv file in python labraries
df=pd.read_csv("phone pay data.csv")
```

```python
# Display first 5 rows
df.head()
```

```python
# Display row and column
df.shape
```

```python
# Display all column name 
df.columns
```

```python
# Check Data types
df.dtypes
```

```python
# Check the data information
df.info()
```

```python
# Convert the transaction date into currect date format
df["Transaction_Date"]=pd.to_datetime(df["Transaction_Date"])
df["Transaction_Date"]
```

```python
# Convert the transaction time into currect time format
df["Transaction_Time"] =pd.to_datetime(df["Transaction_Time"],format="%H:%M:%S").dt.time
df["Transaction_Time"]
```

```python
# Data Cleaning

# Check Missing Value
df.isnull().sum()
```

```python
# Check Duplicate Transactions
df["Transaction_ID"].duplicated().sum()
```

```python
# check status frequently count
df["Status"].value_counts()
```

```python
# Risk Score Range min to max
df["Risk_Score"].min(),df["Risk_Score"].max()
```

```python
# Extracted hour from transaction time 
df["Hour"]=pd.to_datetime(df["Transaction_Time"],format="%H:%M:%S").dt.hour
df[["Transaction_Time","Hour"]]
```

```python
# Extracted day name from the date
df["Day"]=df["Transaction_Date"].dt.day_name()
df[["Transaction_Date","Day"]]
```

```python
# Extracted the month name from the date
df["month"]=df["Transaction_Date"].dt.month_name()
df[["Transaction_Date","month"]]
```

```python
# classified transaction amount into low,medium,and high using numpy
df["Amount_Category"] = np.select(
    [df["Amount_INR"] < 500, df["Amount_INR"] < 2000],
    ["Low", "Medium"],
    default="High"
)

df[["Amount_INR","Amount_Category"]]
```

```python
# Descriptive Statistical analysis of Amount_INR
print("Mean:", np.mean(df["Amount_INR"]))
print("Median:", np.median(df["Amount_INR"]))
print("Std:", np.std(df["Amount_INR"]))
print("Min:", np.min(df["Amount_INR"]))
print("Max:", np.max(df["Amount_INR"]))
```

```python
# check total transaction 
total_transactions = df["Transaction_ID"].nunique()
print("Total_transactions:",total_transactions)
```

```python
# check total amount 
total_amount=df["Amount_INR"].sum()
print("Total_amount:",total_amount)
```

```python
# check total Cashback
total_cashback = df["Cashback_INR"].sum() 
print("Total_cashback:",total_cashback)
```

```python
# Calculate the success of transaction 
success_rate = (df["Status"].eq("Success").mean()*100)
print("success_rate:",success_rate)
```

```python
# Check the fraud transaction count
df["Is_Suspected_Fraud"].value_counts()
```

```python
# Calculate states-wise transaction amount
state_analysis = df.groupby("State") ["Amount_INR"].sum().sort_values(ascending=False) 
print("state_analysis:",state_analysis)
```

```python
# Calculate UPI app-wise transaction amount
upi_analysis = df.groupby("UPI_App")["Amount_INR"].sum().sort_values(ascending=False)
print("upi_analysis:",upi_analysis)
```

```python
# Calculate transaction type-wise amount
transaction_type_analysis = df.groupby("Transaction_Type")["Amount_INR"].sum().sort_values(ascending=False)
print("transaction_type_analysis:",transaction_type_analysis)
```
